```python
import yt_dlp

def download_audio(url):
    """Скачивает только аудио и конвертирует в MP3 (320 kbps)."""
    ydl_opts = {
        'compat_opts': ['allow-unsafe-extract'],
        'format': 'bestaudio/best',
        'outtmpl': '%(title)s.%(ext)s',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '320',
        }],

    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

def download_video(url):
    """Скачивает видео в лучшем качестве со звуком (в MP4)."""
    ydl_opts = {
        'compat_opts': ['allow-unsafe-extract'],
        'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',  # приоритет: MP4 видео + M4A аудио, потом готовый MP4, потом любой лучший
        'outtmpl': '%(title)s.%(ext)s',
        'merge_output_format': 'mp4',
        'postprocessors': [{
            'key': 'FFmpegVideoConvertor',
            'preferedformat': 'mp4',
        }],
    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

if __name__ == "__main__":
    while True:
        video_url = input("Введите ссылку на видео YouTube (или 'q' для выхода): ").strip()
        if video_url.lower() in ('q', 'quit', 'exit'):
            print("Завершение программы.")
            break

        download_video_choice = input("Скачать видео в лучшем качестве со звуком? (y/n): ").strip().lower()
        download_video_flag = download_video_choice in ('y', 'yes', 'д', 'да')

        try:
            if download_video_flag:
                print("Начинаю скачивание видео со звуком...")
                download_video(video_url)
                print("Видео успешно скачано.\n")
            else:
                print("Видео скачивать не будем.\n")
        except Exception as e:
            print(f"Ошибка при скачивании видео: {e}\n")

        download_audio_choice = input("Скачать аудио в лучшем качестве? (y/n): ").strip().lower()
        download_audio_flag = download_audio_choice in ('y', 'yes', 'д', 'да')

        try:
          if download_audio_flag:
            print("Начинаю скачивание аудио...")
            download_audio(video_url)
            print("Аудио успешно скачано.\n")
        except Exception as e:
            print(f"Ошибка при скачивании аудио: {e}\n")

        print("Скачивание завершено.\n")
```
